```python
import pandas as pd

maha = pd.read_excel(
    "../data/raw/Harvested_Extent-Maha_Seasons-1979-2025.xlsx",
    header=None,
)

years = maha.iloc[2].ffill().astype(str).str.strip()
years = years.replace({"2000/2021": "2020/2021"})
categories = maha.iloc[3].astype(str).str.strip()

regions = maha.iloc[4:, 1].astype("string").str.strip().str.upper()
rows = regions.notna() & regions.ne("SRI LANKA")
columns = categories.eq("Total") & years.str.fullmatch(r"\d{4}/\d{4}")

values = maha.iloc[4:].loc[rows, columns].apply(
    pd.to_numeric, errors="coerce"
)
values.columns = years[columns].to_numpy()

maha_by_year = (
    values.sum(min_count=1)
    .rename_axis("year")
    .reset_index(name="harvested_extent_ha")
    .groupby("year", as_index=False)["harvested_extent_ha"]
    .sum(min_count=1)
    .sort_values("year")
)

display(maha_by_year)
```
